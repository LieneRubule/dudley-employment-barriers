---

## Purpose

Clean and validate the Dudley disability dataset for use in later
analysis of local employment barriers and possible support needs.

This notebook covers loading and inspecting the raw data, handling
non-data rows, checking missing values and duplicates, standardising
column names, validating counts and saving a cleaned CSV.

The original dataset will remain unchanged. Analysis and
visualisations will be completed in separate notebooks.

## Ethical Considerations

Disability status does not establish whether someone can work,
wants to work or needs a particular type of support.

The data describes neighbourhood populations. It must not be used
to make assumptions about individual residents.

## Objectives

- Inspect the structure and contents of the raw disability dataset.
- Identify and handle non-data rows, missing values and duplicates.
- Standardise column names and check neighbourhood identifiers.
- Validate numeric counts and data types.
- Document cleaning decisions and save a cleaned CSV.

## Inputs

- Raw dataset: `raw_datasets/dudley_disability.csv`.
- Census 2021 disability data for Dudley, accessed through ONS/Nomis.
- Source headings and notes describing the population and categories.

## Outputs

- Cleaned dataset: `cleaned_datasets/dudley_disability_cleaned.csv`.
- Documented cleaning steps, validation results and a final conclusion.
- Consistent `LSOA_Code` and `LSOA_Name` columns for later matching
  with other datasets.
