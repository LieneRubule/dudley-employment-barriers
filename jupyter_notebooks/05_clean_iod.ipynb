## 1. Inspect the Raw File

Inspect the first ten lines to identify the column headers and any
source information before choosing how to load the dataset.
The original file will remain unchanged.

```python
# Developed with ChatGPT assistance to inspect the raw file structure.
from itertools import islice

raw_path = "raw_datasets/dudley_iod.csv"

# Display the first ten lines with line numbers.
with open(raw_path, "r", encoding="utf-8-sig") as file:
    for line_number, line in enumerate(islice(file, 10), start=1):
        print(f"{line_number}: {line.rstrip()}")
```

### Findings
- Column headers appear on the first line.
- The file includes areas outside Dudley, so filtering is required.
- LSOA codes and names use 2011 geography.
- Local authority columns refer to 2019 boundaries.
- Several population columns refer to mid-2015.
- The deprivation release year needs confirmation from the source;
  geographic dates alone do not establish the release year.
- Geographic compatibility with the Census 2021 datasets must be
  checked before merging.
 
 ---

## 2. Load and Preview the Dataset

Load the CSV using the first row as column headers.
Check the number of rows and columns, then preview five records.
Keep the raw dataframe unchanged for reference.

```python
# Developed with ChatGPT assistance to load and inspect the dataset.
import pandas as pd

# Load the raw CSV without skipping any rows.
iod_df = pd.read_csv(raw_path, encoding="utf-8-sig")

# Check the dataset size and preview five records.
print(f"Rows: {iod_df.shape[0]}")
print(f"Columns: {iod_df.shape[1]}")

display(iod_df.head())
```

### Findings
- The dataset loaded successfully with 32,844 rows and 57 columns.
- The five-row preview shows geographic identifiers, deprivation
  measures and population figures in separate columns.
- Records include City of London and Barking and Dagenham,
  confirming that the file is not restricted to Dudley.
- The raw dataset is retained in `iod_df`; no filtering or
  cleaning has been applied.

---

## 3. Select Dudley Records

Select records where the local authority district name is Dudley.
Create a separate copy so the full raw dataframe remains unchanged.
Check the number of selected records and preview five rows.

```python
# Developed with ChatGPT assistance to select Dudley records.

# Filter by the local authority name and retain all columns.
dudley_mask = (
    iod_df["Local Authority District name (2019)"] == "Dudley"
)
iod_clean = iod_df.loc[dudley_mask].copy().reset_index(drop=True)

# Check the selected dataset size and local authority code.
print(f"Dudley rows: {iod_clean.shape[0]}")
print(f"Columns: {iod_clean.shape[1]}")
print(
    "Local authority codes:",
    iod_clean["Local Authority District code (2019)"].unique()
)

display(iod_clean.head())
```

### Findings
- Filtering selected 201 Dudley records and retained all 57 columns.
- All selected records have local authority code `E08000027`.
- The five-row preview shows Dudley LSOA names.
- The selected data is stored separately in `iod_clean`, leaving
  `iod_df` unchanged.
- The row count differs from the 203 rows in the cleaned Census
  datasets. The LSOA codes and boundary differences need checking
  before merging; this does not yet establish missing data.

---

## 4. Check Missing Values and Duplicates

Check the Dudley subset for missing values, duplicate rows and
repeated LSOA codes. Confirm the number of unique codes before
comparing geographic coverage with the Census datasets.

```python
# Developed with ChatGPT assistance to check data completeness.

# Show only columns containing missing values.
missing_values = iod_clean.isna().sum()
print("Columns with missing values:")
print(missing_values[missing_values > 0].to_string())

# Check duplicate records and geographic identifiers.
print("\nDuplicate rows:", iod_clean.duplicated().sum())
print(
    "Duplicate LSOA codes:",
    iod_clean["LSOA code (2011)"].duplicated().sum()
)
print(
    "Unique LSOA codes:",
    iod_clean["LSOA code (2011)"].nunique()
)
```

### Findings
- No missing values were detected across the 57 columns.
- No duplicate rows or duplicate LSOA codes were found.
- The Dudley subset contains 201 unique LSOA codes.
- No missing-value treatment or duplicate removal is required.
