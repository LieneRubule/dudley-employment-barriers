## 1. Inspect the Raw File

## 2. Load and Preview the Dataset

Load the CSV using the first row as column headers.
Check the number of rows and columns, then preview five records.

  cleaning has been applied.

## 3. Select Dudley Records

Select records where the local authority district name is Dudley.
Create a separate copy so the full raw dataframe remains unchanged.
Check the number of selected records and preview five rows.

```python
# Developed with ChatGPT assistance to select Dudley records.

# Filter by the local authority name and retain all columns.
dudley_mask = (
    iod_df["Local Authority District name (2019)"] == "Dudley"
)
iod_clean = iod_df.loc[dudley_mask].copy().reset_index(drop=True)

# Check the selected dataset size and local authority code.
print(f"Dudley rows: {iod_clean.shape[0]}")
print(f"Columns: {iod_clean.shape[1]}")
print(
    "Local authority codes:",
    iod_clean["Local Authority District code (2019)"].unique()
)

display(iod_clean.head())
```

### Findings
- Filtering selected 201 Dudley records and retained all 57 columns.
- All selected records have local authority code `E08000027`.
- The five-row preview shows Dudley LSOA names.
- The selected data is stored separately in `iod_clean`, leaving
  `iod_df` unchanged.
- The row count differs from the 203 rows in the cleaned Census
  datasets. The LSOA codes and boundary differences need checking
  before merging; this does not yet establish missing data.

## 4. Check Missing Values and Duplicates

Check the Dudley subset for missing values, duplicate rows and

## 5. Compare LSOA Codes with the Census Dataset

Compare the deprivation dataset's 2011 LSOA codes with the codes
in the cleaned economic activity dataset.

Identify shared codes and codes appearing in only one dataset.
This is an initial compatibility check; it does not by itself
confirm that geographic boundaries are identical.

```python
# Developed with ChatGPT assistance to compare geographic coverage.

# Load the cleaned Census economic activity dataset.
economic_df = pd.read_csv(
    "cleaned_datasets/dudley_economic_activity_cleaned.csv"
)

# Compare unique LSOA codes from the two datasets.
iod_codes = set(iod_clean["LSOA code (2011)"])
census_codes = set(economic_df["LSOA_Code"])

print("Shared LSOA codes:", len(iod_codes & census_codes))
print("Codes only in deprivation data:", len(iod_codes - census_codes))
print("Codes only in Census data:", len(census_codes - iod_codes))

# Display the areas requiring further investigation.
print("\nAreas only in deprivation data:")
display(
    iod_clean.loc[
        iod_clean["LSOA code (2011)"].isin(iod_codes - census_codes),
        ["LSOA code (2011)", "LSOA name (2011)"]
    ]
)

print("\nAreas only in Census data:")
display(
    economic_df.loc[
        economic_df["LSOA_Code"].isin(census_codes - iod_codes),
        ["LSOA_Code", "LSOA_Name"]
    ]
)
```

### Findings
- The datasets share 199 LSOA codes.
- Two codes appear only in the deprivation dataset:
  - `E01009881` — Dudley 011C.
  - `E01033187` — Dudley 006G.
- Four codes appear only in the Census dataset:
  - `E01034306` — Dudley 006H.
  - `E01034307` — Dudley 006I.
  - `E01034308` — Dudley 011E.
  - `E01034309` — Dudley 011F.
- A direct inner merge would retain only the 199 shared codes,
  excluding four Census areas.
- The area names suggest possible boundary splits, but an official
  lookup is needed to confirm the relationships.
- Geographic differences require investigation before deciding
  how to combine the datasets. No records have been removed.

Inspect the data type of every column before selecting or renaming
variables. Geographic identifiers should remain text, while

  (`int64`).
- No numeric measure columns were loaded as text, so no data type
  conversions are needed at this stage.
- Numeric ranges still need validation; correct data types alone
  do not confirm that the values are valid.

## 7. Validate Ranks and Deciles

Check that ranks fall between 1 and the number of records in the
full source dataset (32,844), and that deciles fall between 1 and 10.

Use the full dataset size because these rankings compare areas
nationally, rather than only within Dudley. Retain the original
rank and decile values.

```python
# Developed with ChatGPT assistance to validate ranks and deciles.

# Identify rank and decile columns using their existing names.
rank_columns = [
    column for column in iod_clean.columns if " Rank " in column
]
decile_columns = [
    column for column in iod_clean.columns if " Decile " in column
]

# Count values outside the expected ranges in each column.
invalid_ranks = iod_clean[rank_columns].apply(
    lambda column: (~column.between(1, len(iod_df))).sum()
)
invalid_deciles = iod_clean[decile_columns].apply(
    lambda column: (~column.between(1, 10)).sum()
)

print(f"Rank columns checked: {len(rank_columns)}")
print("Out-of-range ranks:")
print(invalid_ranks.to_string())

print(f"\nDecile columns checked: {len(decile_columns)}")
print("Out-of-range deciles:")
print(invalid_deciles.to_string())
```

### Findings
- All 16 rank columns were checked across the 201 Dudley records.
- No ranks fall outside the expected range of 1 to 32,844.
- All 16 decile columns were checked.
- No deciles fall outside the expected range of 1 to 10.
- No corrections were required. Original ranks and deciles have
  been retained without recalculating them for Dudley.

## 8. Validate Explicit Rate Columns

Check that columns labelled `Score (rate)` contain values between
0 and 1, representing proportions.

Apply this check only to explicitly labelled rates. Other deprivation
scores use different scales and must not be treated as proportions.
Keep the original values unchanged.

```python
# Developed with ChatGPT assistance to validate explicit rate columns.

# Select only columns explicitly labelled as rates.
rate_columns = [
    column for column in iod_clean.columns
    if "Score (rate)" in column
]

# Summarise the range and count invalid values for each rate.
rate_checks = pd.DataFrame({
    "Minimum": iod_clean[rate_columns].min(),
    "Maximum": iod_clean[rate_columns].max(),
    "Outside_0_to_1": iod_clean[rate_columns].apply(
        lambda column: (~column.between(0, 1)).sum()
    )
})

print(f"Rate columns checked: {len(rate_columns)}")
display(rate_checks)
```

### Findings
- All four explicitly labelled rate columns contain values between
  0 and 1, with no out-of-range values.
- Observed ranges are:
  - Income: 0.020–0.408.
  - Employment: 0.021–0.302.
  - Income Deprivation Affecting Children (IDACI): 0.014–0.573.
  - Income Deprivation Affecting Older People (IDAOPI): 0.027–0.451.
- No corrections were required. Rates remain in their original
  proportion format.

## 9. Validate Population Counts

Check the five population columns for negative or fractional values.
Population counts should be non-negative whole numbers.

These columns describe different population groups. Do not add them
all together, because the working-age population overlaps other groups.

```python
# Developed with ChatGPT assistance to validate population counts.

# Identify population columns from their existing names.
population_columns = [
    column for column in iod_clean.columns
    if "population" in column.lower()
    or column.startswith("Dependent Children")
]

# Check each column without changing the original values.
population_checks = pd.DataFrame({
    "Minimum": iod_clean[population_columns].min(),
    "Maximum": iod_clean[population_columns].max(),
    "Negative_values": (
        iod_clean[population_columns] < 0
    ).sum(),
    "Fractional_values": (
        iod_clean[population_columns] % 1 != 0
    ).sum()
})

print(f"Population columns checked: {len(population_columns)}")
display(population_checks)
```

### Findings
- All five population columns were checked across the Dudley subset.
- No negative or fractional values were found.
- Total population values range from 1,139 to 3,122 per area.
- All five columns already use the integer data type (`int64`),
  so no conversion or rounding is required.
- The original population counts have been retained unchanged.

## 10. Check Population Totals

Check whether the three non-overlapping age groups—0–15, 16–59
and 60 and over—sum to the stated mid-2015 total population.

Exclude the employment-domain working-age population from this sum
because it overlaps these age groups. Report any differences without
changing the original counts.

```python
# Developed with ChatGPT assistance to check population totals.

# Select the three age groups covering the total population.
age_columns = [
    "Dependent Children aged 0-15: mid 2015 (excluding prisoners)",
    "Population aged 16-59: mid 2015 (excluding prisoners)",
    "Older population aged 60 and over: mid 2015 (excluding prisoners)"
]
total_column = "Total population: mid 2015 (excluding prisoners)"

# Compare the age-group sum with the stated total.
age_total = iod_clean[age_columns].sum(axis=1)
population_difference = age_total - iod_clean[total_column]
population_mismatch = population_difference.ne(0)

print(
    "Rows where age groups do not match the total:",
    population_mismatch.sum()
)

# Preview any mismatches, limiting the output to five rows.
display(
    iod_clean.loc[
        population_mismatch,
        ["LSOA code (2011)", total_column] + age_columns
    ].head()
)
```

### Findings
- The three age groups sum exactly to the stated mid-2015 total
  population for all 201 Dudley records.
- No mismatches were found, so the displayed exceptions table is empty.
- The overlapping employment-domain working-age population was
  excluded from this calculation.
- No corrections were required.
