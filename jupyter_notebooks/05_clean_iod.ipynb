```python
# Developed with ChatGPT assistance to standardise column names.
import re

# Shorten repeated explanations while retaining measurement types.
def clean_column_name(column):
    column = column.replace(
        "Index of Multiple Deprivation (IMD)", "IMD"
    )
    column = column.replace(
        "Income Deprivation Affecting Children Index (IDACI)", "IDACI"
    )
    column = column.replace(
        "Income Deprivation Affecting Older People (IDAOPI)", "IDAOPI"
    )
    column = re.sub(r"\s*\(where 1 is most deprived[^)]*\)", "", column)
    return re.sub(r"[^A-Za-z0-9]+", "_", column).strip("_")


# Prepare and check the names before applying them.
renamed_columns = [
    clean_column_name(column) for column in iod_clean.columns
]
assert len(renamed_columns) == len(set(renamed_columns)), (
    "Duplicate column names detected."
)

# Rename columns without changing any data values.
iod_clean.columns = renamed_columns

print(f"Columns retained: {iod_clean.shape[1]}")
print("Duplicate column names:", iod_clean.columns.duplicated().sum())

for column in iod_clean.columns:
    print(column)
```
