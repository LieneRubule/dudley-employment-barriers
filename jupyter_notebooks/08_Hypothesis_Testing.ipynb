## 3. Calculate and validate the selected inactivity measure

Combine the counts for looking after home or family, long-term
sickness or disability, and other economic inactivity.

Calculate their percentage using `Economic_Total_16_Plus` from the
same source as the denominator.

Before proceeding, check for missing values, negative counts,
non-positive denominators and combined counts exceeding the population
total. Confirm that the resulting percentages fall between 0 and 100.

These checks verify internal consistency, not independent agreement
with the original source. No observations will be removed automatically.

```python
import pandas as pd

# Load the prepared dataset.
df = pd.read_csv(
    "cleaned_datasets/"
    "dudley_combined_prepared.csv"
)

inactivity_components = [
    "Inactive_Home_Or_Family",
    "Inactive_Long_Term_Sick_Or_Disabled",
    "Inactive_Other"
]

denominator = "Economic_Total_16_Plus"
required_columns = (
    inactivity_components + [denominator]
)

# Confirm that all required columns exist.
missing_columns = [
    column
    for column in required_columns
    if column not in df.columns
]

if missing_columns:
    raise ValueError(
        "Required columns are missing: "
        f"{missing_columns}"
    )

# Validate numeric source counts.
counts = df[required_columns].apply(
    pd.to_numeric,
    errors="raise"
)

checks = pd.Series(
    {
        "Rows with missing required values": (
            counts.isna().any(axis=1).sum()
        ),
        "Rows with negative component counts": (
            (counts[inactivity_components] < 0)
            .any(axis=1)
            .sum()
        ),
        "Rows with non-positive denominators": (
            (counts[denominator] <= 0).sum()
        ),
        "Rows with non-integer counts": (
            (counts.mod(1).ne(0) & counts.notna())
            .any(axis=1)
            .sum()
        )
    },
    name="Number_of_Rows"
)

display(checks.to_frame())

if checks.gt(0).any():
    raise ValueError(
        "Review the flagged source counts "
        "before continuing."
    )

df["Selected_Inactive_Count"] = (
    counts[inactivity_components].sum(axis=1)
)

exceeds_total = (
    df["Selected_Inactive_Count"]
    > counts[denominator]
)

print(
    "Combined counts exceeding population total:",
    exceeds_total.sum()
)

if exceeds_total.any():
    raise ValueError(
        "Some combined inactivity counts "
        "exceed the denominator."
    )

df["Selected_Inactive_Pct"] = (
    df["Selected_Inactive_Count"]
    / counts[denominator]
    * 100
)

outside_range = (
    ~df["Selected_Inactive_Pct"].between(0, 100)
)

print(
    "Percentages outside 0–100:",
    outside_range.sum()
)
print("Number of areas retained:", len(df))

if outside_range.any():
    raise ValueError(
        "Review percentages outside the valid range."
    )

display(
    df[
        ["LSOA_Code", "LSOA_Name"]
        + inactivity_components
        + [
            denominator,
            "Selected_Inactive_Count",
            "Selected_Inactive_Pct"
        ]
    ].head()
)

display(
    df["Selected_Inactive_Pct"]
    .describe()
    .rename(
        "Selected inactivity "
        "(% of residents aged 16+)"
    )
    .to_frame()
)
```

### Findings: selected economic inactivity measure

The selected inactivity measure passed all internal validation checks. There were no missing required values, negative component counts, non-positive denominators or non-integer counts. No combined count exceeded its population total, and all calculated percentages fell between 0 and 100. All 199 neighbourhoods were retained.

The mean neighbourhood percentage was **12.36%**, with a median of **10.91%** and a standard deviation of **5.92 percentage points**. Values ranged from **4.06% to 28.60%**, while the middle 50% of neighbourhoods fell between **7.19% and 16.67%**. These figures show variation in the selected forms of inactivity across the included neighbourhoods.

The mean gives each neighbourhood equal weight; it is not the pooled percentage of Dudley residents in these categories.

For example, Dudley 001A recorded 307 residents across the three selected categories out of 1,637 residents aged 16+, giving **18.75%**.

These checks establish internal consistency, not independent source validation. The measure excludes retired residents and economically inactive students from its numerator and must not be interpreted as total economic inactivity.

## 4. H1 — Health limitations and selected economic inactivity

**Question:** Are neighbourhoods with a higher percentage of residents aged 16+ whose activities are limited a lot also associated with a higher selected economic inactivity percentage?

**Null hypothesis (H0):** The population Spearman rank correlation between these two measures is zero.

**Alternative hypothesis (H1):** The population Spearman rank correlation between these two measures is not zero. A positive association is anticipated, but the test is two-sided.

Spearman correlation measures the strength and direction of a monotonic association without requiring a linear relationship or normally distributed variables. Both measures will use all 199 included neighbourhoods.

The correlation coefficient is the main descriptive result. The conventional p-value assumes independent observations; neighbouring areas may be related, and these areas are not a random sample. It will therefore be treated as an exploratory reference rather than definitive evidence about a wider population. Holm adjustment will be applied after all three primary tests, before drawing significance conclusions.

The health indicator overlaps conceptually with the long-term sickness or disability component of the outcome. An association would not establish an independent causal effect of health limitations or describe individual residents.

```python
# Created with ChatGPT assistance and checked against SciPy documentation.
from scipy.stats import spearmanr

# Select the health indicator and inactivity outcome for H1.
health_column = "Limited_A_Lot_16_Plus_Pct"
outcome_column = "Selected_Inactive_Pct"

h1_data = df[[health_column, outcome_column]]

# Check for missing values before testing.
if h1_data.isna().any().any():
    raise ValueError(
        "Missing values found: "
        "review before testing."
    )

# Both measures must vary across neighbourhoods.
if h1_data.nunique().lt(2).any():
    raise ValueError(
        "Both measures must vary "
        "across neighbourhoods."
    )

# Compare neighbourhood rankings using Spearman correlation.
# A two-sided test allows either direction of association.
# Documentation: https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.spearmanr.html
h1_rho, h1_p = spearmanr(
    h1_data[health_column],
    h1_data[outcome_column],
    alternative="two-sided"
)

# Store results for comparison with H2 and H3.
# Keep full precision for the multiple-testing adjustment.
h1_results = pd.DataFrame({
    "Hypothesis": ["H1: Health limitations"],
    "Number_of_Areas": [len(h1_data)],
    "Spearman_Rho": [h1_rho],
    "Unadjusted_P_Value": [h1_p]
})

# Wait for adjusted results before significance conclusions.
display(h1_results)

print(f"Spearman correlation: {h1_rho:.3f}")
print(
    "Unadjusted approximate two-sided p-value: "
    f"{h1_p:.6g}"
)
print(
    "Holm adjustment pending "
    "completion of H2 and H3."
)
```

### H1 findings — Health limitations

Across the 199 included neighbourhoods, Spearman’s correlation was **0.766**, indicating a strong positive association. Neighbourhoods with a higher percentage of residents whose activities are limited a lot generally also had a higher selected economic inactivity percentage.

The unadjusted approximate two-sided p-value was **1.085 × 10⁻³⁹**. Under the test’s assumptions, this provides evidence against a zero rank correlation. However, the p-value is approximate, possible dependence between neighbouring areas has not been accounted for, and Holm adjustment across H1–H3 is still pending.

The observed direction agrees with H1’s anticipated positive association. It does not establish causation or quantify the percentage of inactivity caused by health limitations.

Interpretation also requires caution because long-term sickness or disability is a component of the selected inactivity outcome. This conceptual overlap may contribute to the association. Age composition and other unmeasured neighbourhood characteristics may also influence both measures.

## 5. H2 — Qualifications and selected economic inactivity

**Question:** Do neighbourhoods with a higher percentage of residents aged 16+ with no formal qualifications tend to have higher selected economic inactivity?

**Null hypothesis (H0):** The population Spearman rank correlation between these two measures is zero.

**Alternative hypothesis (H2):** The population Spearman rank correlation between these two measures is not zero. A positive association is anticipated, but the test is two-sided.

Use Spearman correlation to assess the direction and strength of the monotonic association across the 199 included neighbourhoods. As with H1, the approximate p-value is an exploratory reference subject to sample-size and spatial-dependence limitations. Holm adjustment will follow completion of all three tests.

No formal qualifications does not mean no skills or experience. This individual test does not account for health limitations, age composition or other factors that may influence both qualifications and inactivity.

```python
# Created with ChatGPT assistance for the project's H2 analysis.
from scipy.stats import spearmanr

# Select qualifications and inactivity.
qualification_column = "No_Qualifications_Pct"
outcome_column = "Selected_Inactive_Pct"

h2_data = df[
    [qualification_column, outcome_column]
]

# Check for missing values before testing.
if h2_data.isna().any().any():
    raise ValueError(
        "Missing values found: "
        "review before testing."
    )

# Both measures must vary across neighbourhoods.
if h2_data.nunique().lt(2).any():
    raise ValueError(
        "Both measures must vary "
        "across neighbourhoods."
    )

# Compare neighbourhood rankings using Spearman correlation.
# Test both directions without assuming a positive association.
# Documentation: https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.spearmanr.html
h2_rho, h2_p = spearmanr(
    h2_data[qualification_column],
    h2_data[outcome_column],
    alternative="two-sided"
)

# Keep full precision for comparison and Holm adjustment.
h2_results = pd.DataFrame({
    "Hypothesis": ["H2: No formal qualifications"],
    "Number_of_Areas": [len(h2_data)],
    "Spearman_Rho": [h2_rho],
    "Unadjusted_P_Value": [h2_p]
})

# Wait for adjusted results before significance conclusions.
display(h2_results)

print(f"Spearman correlation: {h2_rho:.3f}")
print(
    "Unadjusted approximate two-sided p-value: "
    f"{h2_p:.6g}"
)
print(
    "Holm adjustment pending completion of H3."
)
```

### H2 findings — Qualifications

Across the 199 included neighbourhoods, Spearman’s correlation was **0.836**, indicating a strong positive association between the percentage of residents with no formal qualifications and the selected economic inactivity percentage.

The unadjusted approximate two-sided p-value was **2.741 × 10⁻⁵³**. Under the test’s assumptions, this provides evidence against a zero rank correlation. However, the p-value is approximate, possible dependence between neighbouring areas remains unaccounted for, and Holm adjustment across H1–H3 is still pending.

The observed direction agrees with H2’s anticipated positive association. Its correlation is numerically higher than H1’s value of 0.766, but this does not establish that qualifications have a larger independent effect or that the two correlations differ statistically.

This test does not account for health limitations, age composition or other potential confounding factors. It cannot establish that lacking formal qualifications causes inactivity or that the same individuals are represented by both measures. No formal qualifications should not be interpreted as an absence of skills or experience.

## 6. H3 — Car availability and selected economic inactivity

**Question:** Do neighbourhoods with a higher percentage of households without a car or van tend to have higher selected economic inactivity?

**Null hypothesis (H0):** The population Spearman rank correlation between these two measures is zero.

**Alternative hypothesis (H3):** The population Spearman rank correlation between these two measures is not zero. A positive association is anticipated, but the test is two-sided.

Use Spearman correlation to assess the direction and strength of the monotonic association across the 199 included neighbourhoods. The approximate p-value has the same sample-size and spatial-dependence limitations as H1 and H2. Holm adjustment will follow this test.

Car availability is measured at household level, while the inactivity outcome concerns residents aged 16+. Their association describes neighbourhood patterns, not whether particular inactive residents have access to a vehicle.

Having no car does not necessarily mean poor transport access. Public transport, affordability, distance to jobs and individual travel needs are not directly measured. This test therefore assesses car availability as a limited proxy for potential transport barriers, not transport poverty itself.

```python
# Code developed with ChatGPT assistance using this project's column names.
from scipy.stats import spearmanr

# Select car availability and inactivity.
car_column = "No_Car_Van_Households_Pct"
outcome_column = "Selected_Inactive_Pct"

h3_data = df[[car_column, outcome_column]]

# Check for missing values before testing.
if h3_data.isna().any().any():
    raise ValueError(
        "Missing values found: "
        "review before testing."
    )

# Both measures must vary across neighbourhoods.
if h3_data.nunique().lt(2).any():
    raise ValueError(
        "Both measures must vary "
        "across neighbourhoods."
    )

# Compare rankings using a two-sided Spearman test.
# Reference: SciPy's official spearmanr documentation:
# https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.spearmanr.html
# Used to check inputs, the two-sided option, and returned results.
h3_rho, h3_p = spearmanr(
    h3_data[car_column],
    h3_data[outcome_column],
    alternative="two-sided"
)

# Keep full precision for comparison and Holm adjustment.
h3_results = pd.DataFrame({
    "Hypothesis": [
        "H3: Households without a car or van"
    ],
    "Number_of_Areas": [len(h3_data)],
    "Spearman_Rho": [h3_rho],
    "Unadjusted_P_Value": [h3_p]
})

# Wait for adjusted results before significance conclusions.
display(h3_results)

print(f"Spearman correlation: {h3_rho:.3f}")
print(
    "Unadjusted approximate two-sided p-value: "
    f"{h3_p:.6g}"
)
print(
    "All three primary tests completed; "
    "Holm adjustment is next."
)
```

### H3 findings — Car availability

Across the 199 included neighbourhoods, Spearman’s correlation was **0.877**, indicating a strong positive association. Neighbourhoods with a higher percentage of households without a car or van generally also had higher selected economic inactivity.

The unadjusted approximate two-sided p-value was **1.360 × 10⁻⁶⁴**. Under the test’s assumptions, this provides evidence against a zero rank correlation. However, the p-value is approximate, possible dependence between neighbouring areas remains unaccounted for, and Holm adjustment across H1–H3 is still pending.

The observed direction agrees with H3’s anticipated positive association. This correlation is numerically higher than those for qualifications (0.836) and health limitations (0.766). These differences do not establish that the correlations differ statistically or that car availability has the largest independent effect.

Car availability concerns households, while the inactivity measure concerns residents aged 16+. The result does not establish that inactive individuals lack a vehicle. It also cannot determine whether limited car access contributes to inactivity, inactivity reduces the ability to afford a car, or other factors influence both.

No-car households are not a direct measure of transport poverty. Public transport accessibility, travel costs and distance to employment would be needed to investigate transport barriers more directly.

## 7. Adjust for multiple hypothesis tests

Testing three hypotheses creates multiple opportunities for a false-positive conclusion. Apply the Holm adjustment to the three unadjusted p-values, treating H1–H3 as one family of tests with a significance threshold of 0.05.

The adjustment changes the p-values used for significance decisions; it does not change the Spearman correlation coefficients. With valid input p-values, Holm controls the probability of making at least one false rejection across the test family.

These input p-values are approximate and do not account for possible spatial dependence between neighbourhoods. Holm adjustment does not resolve those limitations, confounding or conceptual overlap between measures. Adjusted decisions therefore remain conditional on the underlying test assumptions.

```python
# Code developed with ChatGPT assistance for the project's three tests.
from statsmodels.stats.multitest import (
    multipletests
)

# Combine all three hypothesis results.
hypothesis_results = pd.concat(
    [h1_results, h2_results, h3_results],
    ignore_index=True
)

# Apply Holm adjustment across H1–H3.
# Use a significance threshold of 0.05.
#
# Reference: statsmodels official multipletests documentation:
# https://www.statsmodels.org/stable/generated/statsmodels.stats.multitest.multipletests.html
# Used to check the method and returned values.
# The last two returned values are not needed here.
reject_h0, adjusted_p, _, _ = multipletests(
    hypothesis_results["Unadjusted_P_Value"],
    alpha=0.05,
    method="holm"
)

# Attach adjusted results in the original order.
hypothesis_results["Holm_Adjusted_P_Value"] = (
    adjusted_p
)
hypothesis_results["Reject_H0_At_0_05"] = (
    reject_h0
)

# Format the display while preserving full precision.
display(
    hypothesis_results.style.format({
        "Spearman_Rho": "{:.3f}",
        "Unadjusted_P_Value": "{:.3e}",
        "Holm_Adjusted_P_Value": "{:.3e}"
    })
)

print(
    "True means reject the zero-correlation null "
    "under the test assumptions."
)
print(
    "Holm adjustment does not correct spatial "
    "dependence or establish causation."
)
```

### Findings — Holm-adjusted hypothesis tests

All three Holm-adjusted p-values were below 0.05. We therefore reject each zero-correlation null hypothesis under the test assumptions. All observed associations were positive, agreeing with the anticipated directions.

| Hypothesis | Spearman correlation | Holm-adjusted p-value | Decision at 0.05 |
|---|---:|---:|---|
| H1: Health limitations | 0.766 | 1.085 × 10⁻³⁹ | Reject H0 |
| H2: No formal qualifications | 0.836 | 5.483 × 10⁻⁵³ | Reject H0 |
| H3: Households without a car or van | 0.877 | 4.080 × 10⁻⁶⁴ | Reject H0 |

Holm adjustment did not change any significance decision. H1’s adjusted p-value remained unchanged because it was the largest of the three original p-values and required no increase in this calculation.

These results show strong neighbourhood-level associations with the selected inactivity outcome. They do not establish independent effects, causal relationships or the proportion of inactivity caused by each barrier. The numerical ordering of correlations does not establish that their differences are statistically significant.

The adjusted p-values retain the limitations of the original approximate tests, including possible spatial dependence. Multiple-testing adjustment does not resolve confounding, the health measure’s conceptual overlap with the outcome, or the limitations of car availability as a transport-poverty proxy.

## Ethical Considerations — Hypothesis Testing

### Purpose and Responsible Use

This notebook investigates associations between health limitations, qualification deficits, household car availability and selected economic inactivity across the 199 included Dudley neighbourhoods.

Hypothesis testing should help assess the evidence for these relationships without assigning blame to residents or presenting statistical associations as proof of causation. The findings should not be used alone to determine individual eligibility for employment support, training or other services.

### Using the Prepared Dataset

The notebook uses the combined prepared dataset so that the tests share a consistent geographic coverage and documented variable definitions.

Previous validation supports confidence that selected source values were preserved during preparation. It does not remove limitations in the original data, geographic exclusions or measurement definitions.

### Defining the Selected-Inactivity Outcome

The code creates `Selected_Inactive_Count` by adding the counts for home or family responsibilities, long-term sickness or disability, and other inactivity reasons.

It then calculates `Selected_Inactive_Pct` by dividing this count by `Economic_Total_16_Plus` and multiplying by 100.

This definition excludes retirement and inactive students from the numerator while retaining all usual residents aged 16 and over in the denominator. The result must therefore be described as selected economic inactivity categories, not total economic inactivity, working-age inactivity or an unemployment rate.

Clear labelling is important because a broader description would imply that the analysis answers questions beyond the available data.

### Respecting Different Reasons for Inactivity

The combined outcome brings together categories that represent different circumstances. Caring responsibilities, long-term health limitations and other reasons for inactivity should not be assumed to have the same causes or require the same response.

Unpaid caring can involve substantial work and social contribution. Economic inactivity should not automatically be presented as a personal failing or a condition that every individual wishes to change.

Combining these categories provides a defined analytical measure, but it also hides differences between them. Interpretation should retain this limitation.

### Validating the Derived Outcome

Checks of missing values, negative counts, population totals and percentage ranges help identify calculation or input problems before testing.

These checks should reveal unexpected values rather than silently replace them. A missing value should not be treated as zero, and a percentage outside its expected range should be investigated rather than clipped automatically.

Passing arithmetic and range checks confirms consistency with the chosen calculation. It does not establish that the outcome captures every relevant form of inactivity or every resident’s circumstances.

### Choosing the Three Hypotheses

The three tests examine selected inactivity against severe activity limitations, no formal qualifications and households without a car or van.

These measures should retain their documented meanings. They are relevant indicators for the research question, but none provides a complete measurement of health-related employment barriers, skills or transport poverty.

The hypotheses were informed by exploratory analysis of the same dataset. They should therefore be presented as exploratory tests rather than independent confirmation of relationships specified before the data was examined.

### Using Spearman Correlation

Spearman correlation assesses whether neighbourhoods with higher values of one measure tend to have higher or lower values of another in a monotonic relationship.

Using ranks can be appropriate for exploring relationships without assuming a straight-line association or normally distributed variables. However, it does not remove the need to consider independence, measurement quality and the population to which the findings apply.

The method does not estimate a causal effect or the change in inactivity expected from changing a predictor by one percentage point.

### Checking Missing and Constant Measures

Checks for missing observations and constant values help establish whether a correlation can be calculated meaningfully.

Any exclusions caused by missing data should be reported because they change the set of areas being compared. Values should not be filled in merely to obtain a result unless there is a justified and documented method.

A constant measure provides no variation for a rank correlation. This should be treated as a limitation of the available comparison rather than a reason to modify the values.

### Using Two-Sided Tests

The tests use two-sided alternatives, allowing evidence of either a positive or negative association. An anticipated positive relationship should not prevent an unexpected negative result from being considered.

The direction and strength of the observed relationship should be reported alongside the p-value. Statistical significance alone does not explain what the association looks like or why it occurs.

### Health Limitations and Conceptual Overlap

The health predictor measures residents aged 16 and over whose day-to-day activities are limited a lot. The outcome includes inactivity attributed to long-term sickness or disability.

These concepts overlap, so their association should not be presented as entirely independent evidence that health limitations cause inactivity. The area-level data also cannot establish whether the same residents are represented in both categories.

Disability should not be interpreted as unwillingness or inability to work. Accessibility, support, job requirements and available opportunities may influence participation but are not directly resolved by this test.

### Formal Qualifications and Skills

The qualification test uses the percentage of residents aged 16 and over with no formal qualifications. This does not measure all practical skills, employment experience, informal learning or capacity to undertake training.

An association with selected inactivity should not be interpreted as evidence that individual residents are inactive because they lack qualifications.

Age composition and other neighbourhood circumstances may influence both measures. The separate correlation does not account for these possible explanations.

### Car Availability and Transport Poverty

The transport-related test uses the percentage of households without a car or van. This is a household measure, while the inactivity outcome is based on people.

The relationship therefore concerns neighbourhood characteristics rather than matched observations of individual transport access and employment circumstances.

No car availability does not necessarily mean poor transport access. Public transport provision, affordability, accessibility, travel times and proximity to employment are not directly measured.

A strong association should not be presented as proof that providing a vehicle would reduce inactivity.

### Source-Specific Denominators

Economic activity, disability and qualification measures retain the population totals from their respective sources. Car availability uses household totals.

These distinctions should remain clear even when the variables are expressed as percentages. A percentage does not make measures with different populations interchangeable.

The population aged 16 and over includes people still studying and people beyond conventional working ages. The tests should not be described as analysing only the working-age population.

### Interpreting Correlation Strength

The correlation coefficient describes the direction and strength of the observed rank association across the included areas.

It does not show how much inactivity is caused by a barrier or the percentage of inactivity explained by that barrier. Squaring a Spearman coefficient should not be presented as a causal contribution.

The three separate correlations do not estimate independent effects. They also do not formally establish that one correlation is statistically stronger than another simply because its numerical value is larger.

### Approximate P-Values

The reported Spearman p-values are approximate. SciPy’s documentation cautions that the approximation is most accurate for very large samples and suggests considering permutation testing for smaller samples.

With 199 areas, the p-values should be described as exploratory and interpreted alongside the coefficients and methodological limitations. No permutation test was performed in this notebook.

A very small p-value does not remove uncertainty about the measurement definitions, geographic coverage or assumptions behind the analysis.

### Spatial Dependence

Neighbouring LSOAs may share transport networks, employment opportunities, housing conditions and other influences. Their observations may therefore not be independent.

Ordinary correlation p-values do not account for this spatial dependence. The apparent amount of independent information may be smaller than the number of areas suggests.

This limitation should remain explicit when interpreting the strength of the statistical evidence.

### Applying the Holm Correction

Testing three hypotheses creates multiple opportunities to identify an apparently significant result. The Holm adjustment addresses this issue across the three reported tests.

The adjustment changes the p-values used for the rejection decisions, not the correlation coefficients. It should not be described as strengthening the relationships.

Holm correction does not repair inaccurate input p-values, spatial dependence, confounding or the selection of hypotheses after exploration. Its protection also does not automatically extend to every relationship previously examined in notebook 07.

### Interpreting Statistical Decisions

Rejecting a null hypothesis indicates evidence against zero population rank correlation under the test’s assumptions. It does not prove the alternative explanation, establish causation or show that the relationship is practically important.

Likewise, failing to reject a null hypothesis would not prove that no relationship or barrier exists.

The significance threshold should be applied consistently rather than changed after inspecting results. Coefficients, p-values, sample sizes and limitations should be considered together.

### Unmeasured Influences and Independent Effects

Age structure, caring responsibilities, local employment opportunities, housing and other circumstances may affect both the predictors and selected inactivity.

The separate tests do not control for these influences or disentangle the relationships among health, qualifications and car availability.

The notebook should therefore not claim to identify the most important independent driver of inactivity. Predictive modelling is outside the current scope and would not, by itself, establish causation either.

The analysis should not attempt to identify individuals or attach sensitive characteristics to particular households.
