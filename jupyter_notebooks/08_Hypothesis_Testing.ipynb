# Hypothesis Testing: Structural Barriers and Economic Inactivity in Dudley

## Main research question

**To what extent do localized structural barriers—specifically health
limitations, qualification deficits, and transport poverty—drive
economic inactivity across Dudley’s 199 neighborhoods?**

## Purpose

Examine how indicators of health limitations, formal qualifications
and car availability relate to economic inactivity across the
199 included Dudley neighbourhoods.

The hypotheses address each of these three dimensions. Their results
will help assess the direction and strength of neighbourhood-level
associations. Where appropriate, a combined model will examine whether
each association remains after accounting for the other selected
indicators.

Although the research question uses the word “drive”, this observational
analysis cannot establish causal effects. Its conclusions will describe
associations and statistical explanatory power, not prove that the
measured characteristics cause economic inactivity.

## Data and coverage

Use `cleaned_datasets/dudley_combined_prepared.csv`, which contains
199 geographically compatible LSOAs and 105 columns.

The dataset combines Census 2021 measures with historical IoD2019
deprivation measures. Four Census areas without matching deprivation
codes were excluded. Findings therefore concern the 199 included
neighbourhoods, rather than all 203 Census areas in Dudley.

## Defining the measures

- **Health limitations:** Use the percentage of residents aged 16
  and over whose day-to-day activities are limited a lot.
- **Qualification deficits:** Use the percentage of residents aged
  16 and over with no formal qualifications. This measures formal
  qualifications, not all skills or experience.
- **Transport poverty:** Use households without a car or van as
  a limited indicator of car availability. It does not directly
  measure transport poverty, affordability or access to public transport.
- **Economic inactivity:** Verify a suitable total economic inactivity
  measure and its denominator before testing. The selected inactivity
  categories currently available must not be labelled as total inactivity.

Do not substitute unemployment or employment shares for economic
inactivity. If a suitable total cannot be obtained, revise the outcome
and hypothesis wording explicitly before proceeding.

## Ethics and interpretation

Economic inactivity is not the same as unemployment and should not
automatically be treated as a problem requiring an individual to work.

Disability does not imply inability or unwillingness to work.
No formal qualifications does not mean no skills.
Having no car does not necessarily mean poor transport access.

The analysis concerns neighbourhoods, not identifiable individuals.
It cannot establish that the people represented by one measure are
the same people represented by another.

Age composition, retirement, caring responsibilities, local employment
opportunities and other unmeasured factors may influence the patterns.
Results must not be used to blame residents, label neighbourhoods
as deficient or justify restricting support.

## Analytical status

These hypotheses are informed by the project aims and the EDA.
Testing them using the same dataset does not provide independent
confirmation.

Report effect sizes and uncertainty alongside any significance tests.
Consider correlated predictors, overlapping definitions, unusual
observations and possible dependence between neighbouring areas.

## 1. Hypotheses and their contribution to the research question

The hypotheses below use a verified economic inactivity percentage
as the outcome. Testing will begin only after its definition,
coverage and denominator have been checked.

### H1: Health limitations and economic inactivity

**Research hypothesis:** Neighbourhoods with a higher percentage
of residents whose day-to-day activities are limited a lot tend
to have a higher economic inactivity percentage.

**H0:** There is no association between the health-limitation
indicator and economic inactivity across the included neighbourhoods.

**H1:** There is an association, with a positive direction anticipated.

**Contribution to the main question:** Assess how strongly the
health-limitation indicator is associated with differences in
economic inactivity between neighbourhoods.

**Ethical and analytical considerations:** An area-level association
cannot establish whether disabled individuals are economically
inactive. Age composition may influence both measures. Total
inactivity also includes inactivity attributed to long-term sickness
or disability, creating conceptual overlap that must be acknowledged.

### H2: Formal qualifications and economic inactivity

**Research hypothesis:** Neighbourhoods with a higher percentage
of residents with no formal qualifications tend to have a higher
economic inactivity percentage.

**H0:** There is no association between the no-qualifications
percentage and economic inactivity across the included neighbourhoods.

**H1:** There is an association, with a positive direction anticipated.

**Contribution to the main question:** Assess whether differences
in formal qualifications are associated with neighbourhood differences
in economic inactivity.

**Ethical and analytical considerations:** Formal qualifications do
not capture practical skills, overseas experience or informal learning.
Age and retirement patterns may contribute to the association.
Results cannot establish that a lack of qualifications causes
inactivity or that training alone would change it.

### H3: Car availability and economic inactivity

**Research hypothesis:** Neighbourhoods with a higher percentage
of households without a car or van tend to have a higher economic
inactivity percentage.

**H0:** There is no association between the no-car household
percentage and economic inactivity across the included neighbourhoods.

**H1:** There is an association, with a positive direction anticipated.

**Contribution to the main question:** Assess whether limited
household car availability is associated with neighbourhood differences
in economic inactivity.

**Ethical and analytical considerations:** Car availability is not
a direct measure of transport poverty. Public transport, accessibility,
travel costs and personal choice are not captured. Household measures
cannot establish the transport circumstances of economically inactive
individuals. The direction of any relationship is also uncertain:
economic circumstances may affect car ownership.

### Bringing the hypotheses together

First examine each relationship separately. If the data and diagnostic
checks support it, examine the three indicators together in a model
of economic inactivity.

This will help distinguish their individual associations from shared
patterns and quantify how much between-neighbourhood variation the
model accounts for. Explained variation must not be interpreted as
the percentage of inactivity caused by these barriers.

### Testing principles

Specify the statistical form of each null hypothesis when selecting
the test or model. Use two-sided tests, report the observed direction
and apply a Holm correction across the three primary hypothesis tests.

Any additional models or sensitivity analyses will be clearly labelled.
Multiple-testing correction does not resolve exploratory selection,
unmeasured confounding or dependence between neighbouring areas.

The final answer will distinguish observed associations, possible
support implications and questions requiring further evidence.
